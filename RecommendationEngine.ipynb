### Text processing using NLP techniques

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words_remove = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
# Creating a function tp preprocess text
def text_processing(text):
    text = str(text).lower()
    # removing uneccesary punctuations
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [word for word in words if word not in stop_words_remove]
    # lemmatize words
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

```python
df['tag'] = df['tag'].apply(text_processing)
```

```python
df = df.reset_index(drop=True)
```

```python
# create indices according to movie titles
indices = pd.Series(df.index,index=df['title'])
```

```python
indices
```

```python
# Vectorize words
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features = 50000, ngram_range = (1,2))
```

```python
tfidf_matrix = tfidf.fit_transform(df['tag'])
```

```python
tfidf_matrix.shape
```

```python
print(tfidf_matrix)
```

```python
# Implement Cosine Similarity for Recommendation
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# No. of movie to be recommended while clicking a title
def recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']
    idx = indices[title]
    sim_score = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]
    return df['title'].iloc[similar_idx]
```

```python
recommend('Toy Story')
```

```python
recommend('Jumanji')
```

```python
#Store model in a pickle file
import pickle

pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))
pickle.dump(indices,open('indices.pkl','wb'))
df.to_pickle('df.pkl')
pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
